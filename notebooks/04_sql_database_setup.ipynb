# SQL Analysis 3: Sales Performance by Order Status

## Business Question

How are orders distributed across different order statuses?

---

## Business Importance

Understanding order status helps businesses monitor operational efficiency and customer satisfaction.

Businesses can identify:

- Successfully delivered orders
- Cancelled orders
- Processing bottlenecks
- Customer fulfillment performance

Monitoring these metrics helps improve logistics, inventory management, and overall customer experience.

---

## SQL queries

- GROUP BY
- COUNT()
- ORDER BY

````python
pd.read_sql(
    """
    SELECT
        order_status,
        COUNT(DISTINCT order_id) AS total_orders
    FROM olist_master
    GROUP BY order_status
    ORDER BY total_orders DESC;
    """,
    connection
)
````

````python
pd.read_sql(
    """
    SELECT
        order_status,
        COUNT(DISTINCT order_id) AS total_orders,
        ROUND(
            COUNT(DISTINCT order_id) * 100.0 /
            (SELECT COUNT(DISTINCT order_id) FROM olist_master),
            2
        ) AS percentage
    FROM olist_master
    GROUP BY order_status
    ORDER BY total_orders DESC;
    """,
    connection
)
````

## 📦 Business Insight

The majority of marketplace orders are expected to be successfully delivered, indicating healthy fulfillment operations.

Monitoring smaller categories such as cancelled, unavailable, or invoiced orders can help identify operational inefficiencies and improve customer satisfaction.

Regularly tracking order status distribution enables businesses to reduce failed deliveries and optimize the order fulfillment process.

# SQL Analysis 6: Customer Segmentation

## Business Question

Which customers contribute the most to the business, and how are customers distributed based on purchase frequency?

### Business Value

Customer segmentation helps businesses:

- Identify VIP customers
- Build loyalty programs
- Increase customer retention
- Personalize marketing campaigns
- Allocate marketing budgets more efficiently

````python
pd.read_sql("""
SELECT
    CASE
        WHEN total_orders = 1 THEN 'One-time Customer'
        WHEN total_orders BETWEEN 2 AND 4 THEN 'Repeat Customer'
        ELSE 'Loyal Customer'
    END AS customer_segment,
    COUNT(*) AS customers,
    ROUND(AVG(total_orders),2) AS avg_orders
FROM (
    SELECT
        customer_unique_id,
        COUNT(DISTINCT order_id) AS total_orders
    FROM olist_master
    GROUP BY customer_unique_id
) customer_orders
GROUP BY customer_segment
ORDER BY customers DESC;
""", connection)
````

## Business Insight

The extremely small proportion of loyal customers suggests that the business experiences high customer churn after the first purchase.

Potential strategies include:

- Loyalty and rewards programs
- Personalized email campaigns
- Product recommendations
- Subscription or membership programs
- Targeted promotions for repeat customers

Improving repeat purchase rates can substantially increase customer lifetime value without proportionally increasing customer acquisition costs.

````python
pd.read_sql("""
SELECT
    product_category_name,
    ROUND(SUM(payment_value), 2) AS total_revenue,
    COUNT(DISTINCT order_id) AS total_orders,
    COUNT(DISTINCT customer_unique_id) AS unique_customers
FROM olist_master
GROUP BY product_category_name
ORDER BY total_revenue DESC
LIMIT 10;
""", connection)
````

The Home, Bed & Bath (cama_mesa_banho) category generated the highest revenue, exceeding 1.74 million across more than 9,400 orders.

Beauty & Health and Computer Accessories closely followed, indicating strong customer demand across both lifestyle and technology products.

The top-performing categories consistently exhibit high revenue, high order volume, and a large unique customer base, suggesting these categories form the company's primary revenue drivers.

## Business Insight

The business should prioritize investment in high-performing categories such as Home, Bed & Bath and Beauty & Health through inventory optimization, supplier partnerships, and targeted marketing campaigns.

Technology-related categories also demonstrate strong customer demand, presenting opportunities for cross-selling and premium product offerings.

Lower-performing categories should be evaluated for profitability and inventory optimization to improve overall operational efficiency.

# SQL Analysis 8: Seller Performance Analysis

## Business Question

Which sellers generate the highest revenue and process the largest number of customer orders?

---

## Business Value

Seller performance analysis helps businesses:

- Identify top-performing sellers
- Build strategic partnerships
- Reward high-performing sellers
- Improve seller onboarding
- Optimize marketplace performance

````python
pd.read_sql("""
SELECT
    seller_id,
    ROUND(SUM(price), 2) AS total_sales,
    COUNT(DISTINCT order_id) AS total_orders,
    ROUND(AVG(price), 2) AS avg_order_value
FROM olist_master
GROUP BY seller_id
ORDER BY total_sales DESC
LIMIT 10;
""", connection)
````

## Observation

The top-performing seller generated total sales of 244,627.55 from only 358 orders, achieving the highest average order value (559.79) among the top sellers.

Other leading sellers processed more than 1,000 orders but recorded lower average order values, indicating a high-volume sales strategy.

These results suggest that marketplace success is driven by both premium pricing and high transaction volume, depending on the seller's business model.

## Business Insight

The marketplace should adopt different engagement strategies for premium and high-volume sellers.

Premium sellers can be supported through exclusive promotions, premium placement, and long-term partnerships, while high-volume sellers may benefit from logistics optimization and inventory support.

Monitoring seller performance enables the business to strengthen relationships with strategic sellers while reducing dependence on any single supplier.

# SQL Analysis 10: Customer Satisfaction Analysis

---

## Business Question

How does customer satisfaction, measured through review scores, impact order volume and revenue?

---

## Business Value

Customer review analysis helps businesses:

- Measure customer satisfaction
- Improve product and seller quality
- Detect operational issues affecting customer experience
- Increase customer retention
- Protect marketplace reputation

---

## SQL Query
```

````python
pd.read_sql("""
SELECT
    CASE
        WHEN total_orders = 1 THEN 'One-time Customer'
        ELSE 'Repeat Customer'
    END AS customer_type,

    COUNT(*) AS customers,

    ROUND(AVG(total_orders), 2) AS avg_orders,

    ROUND(AVG(total_spend), 2) AS avg_customer_value

FROM (

    SELECT
        customer_unique_id,
        COUNT(DISTINCT order_id) AS total_orders,
        SUM(payment_value) AS total_spend

    FROM olist_master
    GROUP BY customer_unique_id

) customer_summary

GROUP BY customer_type
ORDER BY avg_customer_value DESC;
""", connection)
````

## Observation

The customer base is dominated by one-time purchasers, representing over 93,000 customers, while only 2,997 customers made repeat purchases.

Although repeat customers account for a much smaller share of the customer base, they place more than twice as many orders on average and generate an average customer value of 487.14 compared with 205.37 for one-time customers.

This demonstrates that repeat customers contribute significantly more value per customer and represent a critical driver of long-term revenue.

## Business Insight

The analysis highlights a substantial opportunity to improve customer retention.

Increasing the repeat purchase rate through loyalty programs, personalized recommendations, email marketing, and post-purchase engagement could significantly increase customer lifetime value without relying solely on acquiring new customers.

Customer retention should be monitored as a key business KPI alongside revenue growth and customer acquisition to support sustainable long-term profitability.
